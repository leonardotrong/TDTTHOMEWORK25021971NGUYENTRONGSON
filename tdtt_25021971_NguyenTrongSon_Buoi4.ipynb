```python
# Check Prime Number
from math import sqrt

def is_prime(number):
    if number < 2:
        return False
    # Check divisibility from 2 to sqrt(number)
    for i in range(2, int(sqrt(number)) + 1):
        if number % i == 0:
            return False
    return True

print(is_prime(6))
```

```python
# Check Perfect Number
# A perfect number is a positive integer that is equal to the sum of its proper divisors.
def is_perfect_number(n):
    divisors_sum = 0
    for i in range(1, n):
        if n % i == 0:
            divisors_sum += i
    
    return divisors_sum == n

print(is_perfect_number(123))
```

```python
# Factorial Function
def factorial(n):
    if n == 0:
        return 1
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(10))
```

```python
# Simple Calculator String Parser
def calculate_string(expression):
    """Parses a string like '2+3' and returns result."""
    if '+' in expression:
        parts = expression.split('+')
        return '%.2f' % (float(parts[0]) + float(parts[1]))
    elif '-' in expression:
        parts = expression.split('-')
        return '%.2f' % (float(parts[0]) - float(parts[1]))
    elif '*' in expression:
        parts = expression.split('*')
        return '%.2f' % (float(parts[0]) * float(parts[1]))
    elif '/' in expression:
        parts = expression.split('/')
        return '%.2f' % (float(parts[0]) / float(parts[1]))
    return "Invalid"

s = input('Enter expression (e.g. 5+10): ')
print(calculate_string(s))
```

```python
# Isomorphic Strings Check
def is_isomorphic(s, t):
    """Checks if two strings are isomorphic (one-to-one mapping of chars)."""
    if len(s) != len(t):
        return False
        
    map_s_to_t = {}
    map_t_to_s = {}
    
    for char_s, char_t in zip(s, t):
        # Check if mapping conflicts exists
        if (char_s in map_s_to_t and map_s_to_t[char_s] != char_t) or \
           (char_t in map_t_to_s and map_t_to_s[char_t] != char_s):
            return False
            
        map_s_to_t[char_s] = char_t
        map_t_to_s[char_t] = char_s
        
    return True

print(is_isomorphic('egg', 'add'))
```
