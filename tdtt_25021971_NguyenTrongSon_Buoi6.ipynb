```python
# Exercise: Remove Duplicates from List
# Input a list of numbers
input_list = list(map(int, input().split()))
unique_list = []

for number in input_list:
    if number not in unique_list:
        unique_list.append(number)

print(unique_list)
```

```python
# Exercise: Cumulative Sum
nums = list(map(int, input().split()))
if not nums:
    print([])
else:
    cumulative = [nums[0]]
    for i in range(1, len(nums)):
        # Current cumulative = Previous cumulative + current number
        cumulative.append(cumulative[i-1] + nums[i])
    print(cumulative)
```

```python
# Exercise: Count Positives, Negatives, Zeros
stats = {'positives': 0, 'negatives': 0, 'zeros': 0}
numbers = list(map(int, input().split()))

for num in numbers:
    if num > 0:
        stats['positives'] += 1
    elif num < 0:
        stats['negatives'] += 1
    else:
        stats['zeros'] += 1

print(stats)
```

```python
# Exercise: Character/Word Frequency
words = list(input().split())
frequency = {}

for word in words:
    if word not in frequency:
        frequency[word] = 1
    else:
        frequency[word] += 1

print(frequency)
```

```python
# Exercise: Find Pairs summing to K
nums = list(map(int, input().split()))
target_k = int(input())
result_pairs = []

while len(nums) > 0:
    current = nums[0]
    complement = target_k - current
    
    # Check if complement exists
    if complement not in nums:
        nums.remove(current)
    else:
        # Handle special case where number equals complement (e.g. 5 + 5 = 10)
        if current == complement:
            if nums.count(current) > 1:
                nums.remove(current)
                nums.remove(current)
                pair = (current, current)
                if pair not in result_pairs:
                    result_pairs.append(pair)
            else:
                nums.remove(current)
        else:
            # Normal case
            nums.remove(current)
            if complement in nums:
                nums.remove(complement)
                pair = (current, complement)
                # Store sorted tuple to handle (1,4) vs (4,1)
                # logic in original code was slightly specific, preserving original intent:
                if pair not in result_pairs:
                    result_pairs.append(pair)

result_pairs.sort()
print(result_pairs)
```

```python
# Exercise: Matrix Input and Transpose
rows, cols = map(int, input().split())
matrix = []

# Input matrix
for i in range(rows):
    row_values = list(map(int, input().split()))
    matrix.append(row_values)

# Print original matrix
print("Original Matrix:")
for i in range(rows):
    for j in range(cols):
        print(matrix[i][j], end=' ')
    print() # Newline
```
