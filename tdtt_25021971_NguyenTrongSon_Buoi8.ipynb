```python
# Exercise 1: Cylinder Class
class Cylinder:
    """Represents a cylinder geometric shape."""
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def surface_area(self):
        """Calculates the surface area: 2 * pi * r * h"""
        return 2 * 3.14 * self.radius * self.height

    def volume(self):
        """Calculates volume: pi * r^2 * h"""
        return 3.14 * self.radius * self.radius * self.height

# Input radius and height
print('Enter radius and height:')
r, h = map(float, input().split())
my_cylinder = Cylinder(r, h)

# Print results formatted to 2 decimal places
print('%.2f' % my_cylinder.surface_area(), '%.2f' % my_cylinder.volume(), sep=' ')
```

```python
# Exercise 3: Point Class and Quadrant Check
from math import sqrt

class Point:
    """Represents a point in 2D space (x, y)."""
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def check_position(self):
        """Determines the position relative to axes."""
        if self.x == 0 and self.y == 0:
            return 1 # Origin
        elif self.x == 0:
            return 2 # On y-axis
        elif self.y == 0:
            return 3 # On x-axis
        return 0 # In a quadrant

    def distance_to_origin(self):
        """Calculates distance from origin (0,0)."""
        return '%.2f' % sqrt(self.x**2 + self.y**2)

print('Enter x and y:')
x, y = map(float, input().split())
p = Point(x, y)

position = p.check_position()
if position == 1:
    print("M is the origin (Goc toa do)")
elif position == 2:
    print("M is on the y-axis (Truc tung)")
elif position == 3:
    print("M is on the x-axis (Truc hoanh)")

print('Distance to origin:', p.distance_to_origin())
```

```python
# Exercise 5: Shop Management
class Shop:
    def __init__(self):
        # Each item is a list: [name, price]
        self.items = []

    def add_item(self, name, price):
        self.items.append([name, price])

    def remove_item(self, name):
        for item in self.items:
            if item[0] == name:
                self.items.remove(item)
                break

    def has_items(self):
        return len(self.items) > 0

    def total_value(self):
        total = 0
        for item in self.items:
            total += item[1]
        return total

    def display_items(self):
        for item in self.items:
            print(f'{item[0]} {item[1]}')

    def clear_shop(self):
        self.items.clear()

my_shop = Shop()
while True:
    stop = input('Do you want to exit? (yes/no): ')
    if stop.lower() == 'yes':
        break
    
    print('Choose action: 2=Add, 3=Remove, 4=Check Empty, 5=Sum, 6=Display, 7=Clear')
    choice = int(input())
    
    if choice == 2:
        name = input('Name: ')
        price = int(input('Price: '))
        my_shop.add_item(name, price)
    elif choice == 3:
        name = input('Name to remove: ')
        my_shop.remove_item(name)
    elif choice == 4:
        print('Shop has items:', my_shop.has_items())
    elif choice == 5:
        print('Total Value:', my_shop.total_value())
    elif choice == 6:
        my_shop.display_items()
    elif choice == 7:
        my_shop.clear_shop()
```

```python
# Exercise 7: Fraction Class (Phan So)
from math import gcd

class Fraction:
    def __init__(self, num1, den1, num2, den2):
        self.num1 = num1
        self.den1 = den1
        self.num2 = num2
        self.den2 = den2

    def simplify_and_print(self, numerator, denominator):
        """Helper to simplify fraction using GCD and print."""
        common = gcd(numerator, denominator)
        numerator //= common
        denominator //= common
        # Ensure negative sign is usually on numerator or normalized
        if denominator < 0:
            numerator = -numerator
            denominator = -denominator
        print(f'{numerator}/{denominator}')

    def add(self):
        # a/b + c/d = (ad + bc) / bd
        num = self.num1 * self.den2 + self.num2 * self.den1
        den = self.den1 * self.den2
        self.simplify_and_print(num, den)

    def divide(self):
        # (a/b) / (c/d) = (a*d) / (b*c)
        num = self.num1 * self.den2
        den = self.den1 * self.num2
        self.simplify_and_print(num, den)

# Example usage
f = Fraction(2, 4, -1, 4)
print('Division result:')
f.divide()
```

```python
# Exercise 10: Library Management
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def count_by_author(self, author_name):
        count = 0
        for book in self.books:
            if book.author == author_name:
                count += 1
        return count

    def count_by_year(self, year):
        count = 0
        for book in self.books:
            if book.year == year:
                count += 1
        return count

n = int(input('Enter number of operations: '))
lib = Library()
while n > 0:
    operation = input().split()
    command = operation[0]
    
    if command == 'ADD':
        # Format: ADD Title;Author;Year
        details = operation[1].split(';')
        new_book = Book(details[0], details[1], details[2])
        lib.add_book(new_book)
    elif command == 'COUNT':
        print(lib.count_by_author(operation[1]))
    elif command == 'COUNTYEAR':
        print(lib.count_by_year(operation[1]))
    n -= 1
```
