```python
a, b, c = map(int, input().split())
if abs(a - b) < c < (a + b):
    print("Yes")
elif abs(a - c) < b < (a + c):
    print("Yes")
elif abs(b - c) < a < (b + c):
    print("Yes")
else: print("No")
```

```python
a, b, c = map(int, input().split())
if abs(a - b) < c < (a + b) or abs(a - c) < b < (a + c) or abs(b - c) < a < (b + c):
    if a == b or b == c or a == c:
        print("Tam giác cân")
    elif a == b == c:
        print("Tam giác đều")
    else:
        print("Tam giác thường")
else: print("Không phải tam giác")
```

```python
n = float(input())
if 0 <= n <= 50:
    print(n * 1500)
elif 51 <= n <= 100:
    print(50 * 1500 + (n - 50) * 2000)
else: 
    print(50 * 1500 + 50 * 2000 + (n - 100) * 3000)
```

```python
n = float(input())
up = (n // 5) + 1
down = n // 5
if abs(n - up) > abs(n - down):
    ans = down
elif abs(n - up) <= abs(n - down):
    ans = up
print(up, down, ans)
```
