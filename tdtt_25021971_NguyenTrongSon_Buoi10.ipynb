```python
# W9A1: Merge two sorted lists without using sort
def merge_sorted_lists(list1, list2):
    """
    Merges two sorted lists into one sorted list using two pointers.
    Time Complexity: O(n + m)
    """
    result = []
    i, j = 0, 0
    n, m = len(list1), len(list2)
    
    # Compare elements from both lists and append the smaller one
    while i < n and j < m:
        if list1[i] < list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1
            
    # Append any remaining elements
    result.extend(list1[i:])
    result.extend(list2[j:])
    
    return result

```

```python
# W9A2: Find k-th largest element in a sorted matrix
def kth_largest_in_matrix(matrix, k):
    """
    Finds the k-th largest element.
    Approach: Flatten the matrix and sort. 
    For optimal performance on massive matrices, a Min-Heap or Binary Search on value range would be used.
    """
    # Flatten the matrix into a single list
    flattened = [elem for row in matrix for elem in row]
    
    # Sort in descending order to easily find k-th largest
    flattened.sort(reverse=True)
    
    return flattened[k-1]

```

```python
# W9A3: Swap 2 digits to minimize the integer
def minimize_number(num_str):
    """
    Swaps two digits to create the smallest possible number.
    Strategy: Greedy approach finding the first non-optimal digit.
    """
    s = list(str(num_str))
    n = len(s)
    
    for i in range(n):
        # Find the smallest digit in the suffix s[i+1:]
        if i == n - 1: break
        
        suffix = s[i+1:]
        min_val = min(suffix)
        
        # If the current digit is larger than the minimum found later
        if s[i] > min_val:
            # Find the LAST occurrence of min_val in the suffix to swap
            # (Swapping with the last occurrence pushes the larger s[i] further back)
            for j in range(n - 1, i, -1):
                if s[j] == min_val:
                    s[i], s[j] = s[j], s[i]
                    return "".join(s)
                    
    return "".join(s)

```

```python
# W9A4: Check if string A can be transformed into B
def can_transform(start_str, target_str):
    """
    Checks if start_str can become target_str by capitalizing and deleting characters.
    """
    i, j = 0, 0
    n, m = len(start_str), len(target_str)
    
    while i < n and j < m:
        if start_str[i] == target_str[j]:
            i += 1
            j += 1
        elif start_str[i].upper() == target_str[j]:
            i += 1
            j += 1
        elif start_str[i].islower():
            # Can delete lowercase characters
            i += 1
        else:
            # Mismatch uppercase character, cannot delete
            return "NO"
            
    # If target is not fully matched
    if j < m:
        return "NO"
        
    # Check if remaining characters in start_str can be deleted (must be lowercase)
    while i < n:
        if start_str[i].isupper():
            return "NO"
        i += 1
        
    return "YES"

```

```python
# W9A5: Check if list1 is a contiguous sublist of list2
def is_sublist(list1, list2):
    """
    Checks if list1 appears contiguously inside list2.
    """
    n1, n2 = len(list1), len(list2)
    if n1 > n2:
        return "NO"
    
    for i in range(n2 - n1 + 1):
        # Check slice equality
        if list2[i : i + n1] == list1:
            return "YES"
            
    return "NO"

```
