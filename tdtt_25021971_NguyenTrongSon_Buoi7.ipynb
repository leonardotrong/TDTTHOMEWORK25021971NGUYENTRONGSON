```python
# Binary Search Algorithm
def binary_search(array, left, right, value):
    """Recursive binary search."""
    if right < left:
        return -1
        
    mid = (left + right) // 2
    
    if array[mid] == value:
        return mid
    elif array[mid] < value:
        # Search right half
        return binary_search(array, mid + 1, right, value)
    else:
        # Search left half
        return binary_search(array, left, mid - 1, value)

sorted_list = [int(n) for n in input('Enter sorted list: ').split()]
target = int(input('Enter target: '))
print(binary_search(sorted_list, 0, len(sorted_list) - 1, target))
```

```python
# Sorting Algorithms: Bubble Sort and Merge Sort
def bubble_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n - 1):
        # Last i elements are already in place
        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                # Swap if the element found is greater than the next element
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

def merge(left_half, right_half):
    result = []
    i = j = 0
    
    # Merge two sorted arrays
    while i < len(left_half) and j < len(right_half):
        if left_half[i] < right_half[j]:
            result.append(left_half[i])
            i += 1
        else:
            result.append(right_half[j])
            j += 1
    
    # Append remaining elements
    result.extend(left_half[i:])
    result.extend(right_half[j:])
    return result

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])
    
    return merge(left_half, right_half)

data = [int(n) for n in input('Enter list: ').split()]
# Note: bubble_sort sorts in-place
print("Bubble Sorted:", bubble_sort(data.copy()))
print("Merge Sorted:", merge_sort(data))
```

```python
# String Search Loop
def check_substring(source, target, start_index):
    """Checks if target string exists in source starting at start_index."""
    i = start_index
    j = 0
    while j < len(target):
        if i >= len(source) or source[i] != target[j]:
            return 0
        i += 1
        j += 1
    return 1

source_str = str(input('Enter string: '))
target_str = str(input('Enter target tag: '))
count = 0

for i in range(len(source_str)):
    if source_str[i] == target_str[0]:
        count += check_substring(source_str, target_str, i)
print(count)
```
